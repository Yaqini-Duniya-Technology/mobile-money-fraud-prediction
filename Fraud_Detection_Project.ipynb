# 🛠 Feature Engineering

---

# 📖 Overview

Feature engineering is the process of creating new variables from existing features to improve the predictive performance of machine learning models.

In fraud detection, engineered features can reveal hidden transaction patterns that are difficult to detect using the original variables alone.

Rather than relying solely on the raw dataset, we derive additional features that capture changes in account balances, transaction proportions, and unusually large transactions.

---

# 🎯 Objectives

In this section, we will:

- Create balance difference features.
- Calculate balance error features.
- Create amount-to-balance ratio features.
- Identify unusually large transactions.
- Validate the newly engineered features.

```python
# Sender Balance Change
# Difference between sender's balance before and after the transaction
df["sender_balance_change"] = (df["oldbalanceOrg"] - df["newbalanceOrig"])

# Receiver Balance Change
# Difference between receiver's balance after and before the transaction
df["receiver_balance_change"] = (
    df["newbalanceDest"] - df["oldbalanceDest"])

# Balance Error (Sender)
# Difference between expected and actual sender balance
df["sender_balance_error"] = (
    df["oldbalanceOrg"] - df["amount"] - df["newbalanceOrig"])

# Balance Error (Receiver)
# Difference between expected and actual receiver balance
df["receiver_balance_error"] = (
    df["oldbalanceDest"] + df["amount"] - df["newbalanceDest"])
```

```python
## Amount-to-Balance Ratio: To avoid division by zero, replace zero balances with NaN during the calculation.

df["amount_balance_ratio"] = (
    df["amount"] /
    df["oldbalanceOrg"].replace(0, np.nan))

df["amount_balance_ratio"] = df["amount_balance_ratio"].fillna(0)


# Large Transaction Indicator :Transactions above the 95th percentile are flagged as large transactions.

threshold = df["amount"].quantile(0.95)

df["large_transaction"] = (df["amount"] > threshold).astype(int)

# Verify the New Features
new_features = [
    "sender_balance_change",
    "receiver_balance_change",
    "sender_balance_error",
    "receiver_balance_error",
    "amount_balance_ratio",
    "large_transaction"]

df[new_features].head()
print("Dataset Shape:", df.shape)
df.columns.tolist()
```

# ⚖️ Handling Class Imbalance (SMOTE)

---

# 📖 Overview

In real-world fraud detection datasets, fraudulent transactions occur much less frequently than legitimate transactions. This creates a **class imbalance problem**, where machine learning models may become biased toward the majority class.

The PaySim dataset is highly imbalanced, with legitimate transactions vastly outnumbering fraudulent ones. If left unaddressed, a model could achieve high accuracy simply by predicting every transaction as legitimate while failing to detect actual fraud.

To overcome this challenge, we use **SMOTE (Synthetic Minority Over-sampling Technique)**.

---

# 🎯 Objectives

In this section, we will:

- Examine the class distribution before SMOTE.
- Apply SMOTE to the training dataset only.
- Generate synthetic fraud samples.
- Verify the new class distribution.
- Visualize the class balance before and after SMOTE.

---

# 📌 Why SMOTE?

SMOTE helps to:

- Balance the dataset.
- Improve fraud detection.
- Reduce model bias toward the majority class.
- Increase Recall and F1-score.
- Improve the model's ability to identify rare fraud cases.

```python
# Display class distribution before SMOTE
print("Class Distribution Before SMOTE:")
print(y_train.value_counts())


# Visualize Class Distribution Before SMOTE
plt.figure(figsize=(6,5))
sns.countplot(x=y_train,palette="Set2")
plt.title("Class Distribution Before SMOTE")
plt.xlabel("Fraud Status")
plt.ylabel("Number of Transactions")
plt.xticks([0,1], ["Legitimate","Fraud"])
plt.show()

# Initialize SMOTE
smote = SMOTE(random_state=42)

# Apply SMOTE: Apply SMOTE only on the training data
X_train_smote, y_train_smote = smote.fit_resample( X_train_scaled,y_train)

# Verify the New Dataset Shapes
print("Original Training Shape :", X_train_scaled.shape)
print("Balanced Training Shape:", X_train_smote.shape)


# Check Class Distribution After SMOTE
print("Class Distribution After SMOTE:")
print(y_train_smote.value_counts())


# Visualize Class Distribution After SMOTE
plt.figure(figsize=(6,5))
sns.countplot( x=y_train_smote, palette="viridis")
plt.title("Class Distribution After SMOTE")
plt.xlabel("Fraud Status")
plt.ylabel("Number of Transactions")
plt.xticks([0,1], ["Legitimate","Fraud"])
plt.show()


# Compare Before and After SMOTE
fig, axes = plt.subplots(1, 2, figsize=(12,5))
sns.countplot(x=y_train,ax=axes[0], palette="Set2")
axes[0].set_title("Before SMOTE")
axes[0].set_xticklabels(["Legitimate","Fraud"])
sns.countplot(x=y_train_smote, ax=axes[1],palette="viridis")
axes[1].set_title("After SMOTE")
axes[1].set_xticklabels(["Legitimate","Fraud"])
plt.tight_layout()
plt.show()

# Convert the balanced training features back to a DataFrame
X_train_smote = pd.DataFrame(X_train_smote, columns=X_train.columns)

# Convert the balanced target to a Series
y_train_smote = pd.Series(y_train_smote,name="isFraud")

# Preview the Balanced Dataset
X_train_smote.head()
y_train_smote.head()
```

# 📈 Model Evaluation

---

# 📖 Overview

Model evaluation is the process of measuring how well a trained machine learning model performs on unseen data.

Since fraud detection is a binary classification problem with highly imbalanced classes, relying solely on accuracy can be misleading.

Instead, we evaluate each model using several complementary metrics to obtain a comprehensive understanding of its performance.

---

# 🎯 Objectives

In this section, we will:

- Evaluate each trained model.
- Generate confusion matrices.
- Display classification reports.
- Plot ROC curves.
- Plot Precision-Recall curves.
- Compare model performance.

---

# 📌 Evaluation Metrics

### Accuracy

Measures the proportion of correctly classified transactions.

---

### Precision

Measures how many transactions predicted as fraud are actually fraudulent.

---

### Recall

Measures how many actual fraudulent transactions are correctly detected.

> **Recall is the most important metric in fraud detection because missing a fraudulent transaction can result in financial loss.**

---

### F1-Score

The harmonic mean of Precision and Recall.

---

### ROC-AUC

Measures the model's ability to distinguish between fraudulent and legitimate transactions.

```python
## Import Additional Metrics
from sklearn.metrics import (ConfusionMatrixDisplay, RocCurveDisplay, PrecisionRecallDisplay)

## Create Evaluation Function
def evaluate_model(model, model_name):
    """
    Evaluate a trained machine learning model.
    """

    # Predictions
    y_pred = model.predict(X_test_scaled)
    y_prob = model.predict_proba(X_test_scaled)[:, 1]

    print("="*70)
    print(f"{model_name}")
    print("="*70)

    # Metrics
    print(f"Accuracy : {accuracy_score(y_test, y_pred):.4f}")
    print(f"Precision: {precision_score(y_test, y_pred):.4f}")
    print(f"Recall   : {recall_score(y_test, y_pred):.4f}")
    print(f"F1 Score : {f1_score(y_test, y_pred):.4f}")
    print(f"ROC AUC  : {roc_auc_score(y_test, y_prob):.4f}")

    print("\nClassification Report\n")
    print(classification_report(y_test, y_pred))

    # Confusion Matrix
    ConfusionMatrixDisplay.from_predictions(y_test, y_pred, cmap="Blues")
    plt.title(f"{model_name} - Confusion Matrix")
    plt.show()

    # ROC Curve
    RocCurveDisplay.from_predictions(y_test, y_prob)
    plt.title(f"{model_name} - ROC Curve")
    plt.show()

    # Precision-Recall Curve
    PrecisionRecallDisplay.from_predictions(y_test, y_prob)
    plt.title(f"{model_name} - Precision-Recall Curve")

    plt.show()
```

```python
## Evaluate Every Model
for model_name, model in models.items():
    evaluate_model(model, model_name)

## Display Summary Table
results_df

## Sort by Recall: Since fraud detection prioritizes finding fraudulent transactions, let's rank the models by Recall.

results_df.sort_values(by="Recall", ascending=False)

## Sort by ROC-AUC
results_df.sort_values(by="ROC-AUC", ascending=False)

## Sort by F1-Score
results_df.sort_values(by="F1 Score", ascending=False)
```

---

# 📌 Why RandomizedSearchCV?

For large datasets like PaySim, **RandomizedSearchCV** is preferred over **GridSearchCV** because it:

- Requires less computational time.
- Explores a wide range of parameter combinations.
- Produces competitive results.
- Scales well to large datasets.

```python

# 1. Downsample base training data to 10% to fit into your RAM comfortably
X_train_small = pd.DataFrame(X_train).sample(frac=0.10, random_state=42)
y_train_small = pd.Series(y_train).loc[X_train_small.index]

# 2. Re-run SMOTE step ONLY on this smaller 10% sample
# (Replace 'smote' with whatever your actual SMOTE object variable is named)
X_train_smote, y_train_smote = smote.fit_resample(X_train_small, y_train_small)

# 3. Force your data to lower precision to save an extra 50% memory
X_train_smote = X_train_smote.astype('float32')
```

```python
## Convert the entire DataFrame to float32
df = df.astype('float32')

# Import Hyperparameter Tuning Tools
from sklearn.model_selection import RandomizedSearchCV


## Define the Random Forest Hyperparameter Grid
rf_params = {"n_estimators": [100, 200, 300], "max_depth": [10, 20, 30, None], "min_samples_split": [2, 5, 10], "min_samples_leaf": [1, 2, 4], "max_features": ["sqrt", "log2"]}

## Initialize the Random Forest Model
rf = RandomForestClassifier(random_state=42, n_jobs=-1)
```

```python
from joblib import parallel_backend
## Configure RandomizedSearchCV:  We use Recall as the optimization metric because detecting fraudulent transactions is our primary objective.
rf_random = RandomizedSearchCV( estimator=rf, param_distributions=rf_params, n_iter=5, scoring="recall", cv=3, verbose=3, random_state=42, n_jobs=1)


## Train the Search using the threading backend to save RAM
with parallel_backend('threading'):
    rf_random.fit(X_train_smote, y_train_smote)

## Display the Best Parameters
print("Best Parameters:")
rf_random.best_params_


## Display the Best Recall Score
print("Best Cross-Validation Recall:")
print(rf_random.best_score_)


## Retrieve the Best Model
best_rf = rf_random.best_estimator_


## Make Predictions
y_pred_rf = best_rf.predict(X_test_scaled)
y_prob_rf = best_rf.predict_proba(X_test_scaled)[:, 1]


## Evaluate the Tuned Model
print("Accuracy :", accuracy_score(y_test, y_pred_rf))
print("Precision:", precision_score(y_test, y_pred_rf))
print("Recall   :", recall_score(y_test, y_pred_rf))
print("F1 Score :", f1_score(y_test, y_pred_rf))
print("ROC-AUC  :", roc_auc_score(y_test, y_prob_rf))


##Classification Report
print(classification_report(y_test, y_pred_rf))


## Confusion Matrix
ConfusionMatrixDisplay.from_predictions(y_test, y_pred_rf, cmap="Greens")
plt.title("Tuned Random Forest - Confusion Matrix")
plt.show()


## ROC Curve
RocCurveDisplay.from_predictions(y_test, y_prob_rf)
plt.title("Tuned Random Forest - ROC Curve")
plt.show()


## Precision-Recall Curve
PrecisionRecallDisplay.from_predictions(y_test, y_prob_rf)
plt.title("Tuned Random Forest - Precision-Recall Curve")
plt.show()
```

```python
## Create Prediction Function
def predict_fraud(transaction):

    """
    Predict whether a transaction is fraudulent.

    Input:
    transaction - pandas DataFrame containing one transaction

    Output:
    prediction and fraud probability
    """

    # Apply same scaling used during training
    transaction_scaled = scaler.transform(transaction)

    # Generate prediction
    prediction = final_model.predict(transaction_scaled)


    # Generate probability
    probability = final_model.predict_proba(transaction_scaled)[:,1]

    if prediction[0] == 1:
        result = "Fraudulent Transaction"
    else:
        result = "Legitimate Transaction"
    return {"Prediction": result, "Fraud Probability": round(probability[0], 4)}

##Create Sample Transaction 
# Select one legitimate transaction example
sample_transaction = X_test.iloc[[0]]
sample_transaction
```
