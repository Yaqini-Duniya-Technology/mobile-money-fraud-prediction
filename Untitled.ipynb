```python
%%writefile core/users.py
# ==========================================
# FraudGuard AI
# User Configuration & RBAC
# ==========================================

USERS = {
    "admin": {
        "password_hash": "$2b$12$7j8HoUoMBaWC4jety7ROluavqgd49yov/mCjP.331.YzJK1wgrGNa",
        "role": "admin"
    },
    "analyst": {
        "password_hash": "$2b$12$RgvYqdihwKarD304LIOuwuQV4eHDwYSsKa6Ly45DZ76LCH539EPWa",
        "role": "analyst"
    },
    "user": {
        "password_hash": "$2b$12$9nSmlmR9dOnbFa.am5EWaejEX4R5Q138TFingb8KodNo9RFZFb3A6",
        "role": "user"
    }
}

from core.security import verify_password

def authenticate_user(username: str, password: str):
    user = USERS.get(username)
    if user is None:
        return None

    valid_password = verify_password(password, user["password_hash"])
    if not valid_password:
        return None

    return {
        "username": username,
        "role": user["role"]
    }
```

```python
import bcrypt

# Passwords for your three users
passwords = {
    "admin": "admin123",
    "analyst": "analyst123",
    "user": "user123"
}

print("\nFraudGuard AI - Password Hash Generator")
print("=" * 50)

for username, password in passwords.items():
    hashed = bcrypt.hashpw(
        password.encode("utf-8"),
        bcrypt.gensalt()
    ).decode("utf-8")

    print(f"\nUsername: {username}")
    print(f"Password: {password}")
    print(f"Hash: {hashed}")

print("\n" + "=" * 50)
print("Copy the hashes into your users.py file.")
```
